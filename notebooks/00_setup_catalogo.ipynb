# 00 - Setup do catálogo
Cria o catálogo do MVP, os schemas da arquitetura medalhão (bronze, silver, gold) e o volume de arquivos brutos.

```python
%sql

CREATE CATALOG IF NOT EXISTS mvp_pipeline_vf
  COMMENT 'MVP Pipeline de Dados na Nuvem: incentivos fiscais, MPE e crescimento econômico no RS';
```

```python
%sql

USE CATALOG mvp_pipeline_vf;
```

## 1. Schemas da arquitetura medalhão


```python
%sql

CREATE SCHEMA IF NOT EXISTS bronze
  COMMENT 'Dados brutos como recebidos da fonte, acrescidos de metadados de ingestão';

CREATE SCHEMA IF NOT EXISTS silver
  COMMENT 'Dados limpos, tipados, deduplicados e padronizados';

CREATE SCHEMA IF NOT EXISTS gold
  COMMENT 'Modelo dimensional (fatos e dimensões) para responder às perguntas de negócio';
```

## 2. Volume de arquivos brutos


```python
%sql

CREATE VOLUME IF NOT EXISTS bronze.arquivos_brutos
  COMMENT 'Arquivos originais das fontes: SEFAZ-RS, Receita Federal (CNPJ) e IBGE';
```

## 3. Verificação

```python
%sql

SHOW SCHEMAS IN mvp_pipeline_vf;
```
